## Optional: run the full Streamlit app from Colab
This starts the app and opens a free, temporary Cloudflare tunnel. Click the printed `trycloudflare.com` link. Stop the cell to shut it down.

```python
!wget -q -nc https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O cloudflared && chmod +x cloudflared
import subprocess, re
subprocess.Popen(['streamlit', 'run', 'demos/03_dashboard/app.py', '--server.port', '8501', '--server.headless', 'true'])
tunnel = subprocess.Popen(['./cloudflared', 'tunnel', '--url', 'http://localhost:8501'], stderr=subprocess.PIPE, text=True)
for line in tunnel.stderr:
    m = re.search(r'https://[-a-z0-9]+\.trycloudflare\.com', line)
    if m:
        print('👉 Open your dashboard:', m.group(0))
        break
```
